# Введение

**Вероятностным пространством называется тройка:**

$$
(\Omega,\mathcal F,\mathbb P)
$$

где:

- $\Omega$ — множество элементарных исходов (пространство исходов);
- $\mathcal F$ — $\sigma$-алгебра подмножеств $\Omega$, элементы которой называются событиями;
- $\mathbb P:\mathcal F\to[0,1]$ — вероятностная мера, то есть $\sigma$-аддитивная функция на $\mathcal F$, удовлетворяющая $\mathbb P(\Omega)=1$.

**Вопросы:**

1. Что такое сигма-алгебра?
2. Что такое сигма-аддитивная функция на сигма-алгебре?
3. Как задаётся пространство событий и вероятность в конкретных задачах?


# Сигма-алгебра

**Определение:**

Пусть $\Omega$ — непустое множество. Семейство $\mathcal F$ подмножеств множества $\Omega$ называется **$\sigma$-алгеброй** (или **сигма-алгеброй**), если выполнены следующие три условия.

1. **Содержит всё пространство:**

   $$
   \Omega\in\mathcal F
   $$

   Тогда и пустое множество $\varnothing\in\mathcal F$, так как $\varnothing=\Omega\setminus\Omega$.

2. **Замкнутость относительно дополнения:**

   Если $A\in\mathcal F$, то

   $$
   \Omega\setminus A\in\mathcal F
   $$

3. **Замкнутость относительно счётных объединений:**

   Если $A_1,A_2,\ldots,A_n,\ldots$ — счётная последовательность множеств из $\mathcal F$, то

   $$
   \bigcup_{n=1}^{\infty}A_n\in\mathcal F
   $$

**Вопрос: почему** $\mathcal F\ne 2^\Omega$?


# Задача о встрече

**Условие:** два человека договорились встретиться в определённом месте между 12:00 и 13:00. Каждый из них приходит в случайный момент времени в течение этого часа и ждёт другого ровно 20 минут (или $\frac13$ часа). Какова вероятность того, что они встретятся?

**Решение (геометрическое)**

1. **Обозначим:**
   - $x$ — время прихода первого человека (в минутах от 12:00, от 0 до 60);
   - $y$ — время прихода второго человека (от 0 до 60).

2. **Пространство всех исходов** — это квадрат $[0,60]\times[0,60]$ на плоскости. Его площадь:

   $$
   S_{\text{общ}}=60\cdot60=3600
   $$

3. **Условие встречи:** разница во времени прихода не должна превышать 20 минут (ведь первый ждёт второго 20 минут, и второй ждёт первого 20 минут):

   $$
   |x-y|\le20
   $$

4. **Изобразим это на графике:**
   - линии $y=x+20$ и $y=x-20$ отсекают внутри квадрата полосу;
   - встреча происходит в закрашенной области между этими прямыми.

5. **Площадь благоприятной области** проще всего найти, вычтя из площади квадрата площади двух незакрашенных треугольников (где $x-y>20$ или $y-x>20$):
   - каждый треугольник — прямоугольный с катетами $60-20=40$ минут;
   - площадь одного треугольника: $\frac12\cdot40\cdot40=800$;
   - площадь двух треугольников: $1600$.

   Тогда площадь благоприятной области:

   $$
   S_{\text{встр}}=3600-1600=2000
   $$

6. **Вероятность:**

   $$
   \mathbb P=\frac{S_{\text{встр}}}{S_{\text{общ}}}
   =\frac{2000}{3600}
   =\frac59
   \approx0.5556
   $$

**Ответ:** $\frac59$ (примерно 55,56%).

**Проблема:** а для любой фигуры будет площадь?


# Как задаётся вероятность?

**Теорема (Каратеодори)**

Пусть:

- $\Omega$ — непустое множество;
- $\mathcal A$ — **алгебра** подмножеств $\Omega$ (то есть замкнута относительно конечных объединений и дополнений, но **не обязательно** счётных);
- $\mu_0:\mathcal A\to[0,+\infty]$ — **мера** на этой алгебре, то есть:
  - $\mu_0(\varnothing)=0$;
  - $\mu_0$ **счётно-аддитивна** на $\mathcal A$: если $A_1,A_2,\ldots$ — попарно непересекающиеся множества из $\mathcal A$, и их объединение $\bigcup_{n=1}^{\infty}A_n$ также принадлежит $\mathcal A$, то

    $$
    \mu_0\left(\bigcup_{n=1}^{\infty}A_n\right)= \sum_{n=1}^{\infty}\mu_0(A_n)
    $$

Тогда существует мера $\mu$ на $\sigma$-алгебре $\sigma(\mathcal A)$ (минимальной $\sigma$-алгебре, порождённой $\mathcal A$), такая что

$$
\mu(A)=\mu_0(A)
$$

для всех $A\in\mathcal A$.

То есть $\mu$ продолжает $\mu_0$.

Более того, если $\mu_0$ $\sigma$-конечна (то есть $\Omega$ можно представить как счётное объединение множеств из $\mathcal A$ конечной меры), то такое продолжение **единственно**.

**Вывод:** достаточно задать меру (вероятность) на алгебре, порождающей данную сигма-алгебру.

**Определение**

Непустое семейство $\mathcal P$ подмножеств множества $\Omega$ называется **полуалгеброй** (или **полукольцом**), если выполнены три условия.

1. **Пустое множество принадлежит $\mathcal P$:**

   $$
   \varnothing\in\mathcal P
   $$

2. **Замкнутость относительно пересечения:**

   Если $A,B\in\mathcal P$, то их пересечение также принадлежит $\mathcal P$:

   $$
   A\cap B\in\mathcal P
   $$

3. **Свойство разности (дополнения в виде конечной дизъюнктной суммы):**

   Если $A,B\in\mathcal P$ и $A\subseteq B$, то разность $B\setminus A$ можно представить как конечное объединение попарно непересекающихся множеств из $\mathcal P$:

   $$
   B\setminus A=\bigcup_{i=1}^{n}C_i
   $$

   где $C_i\in\mathcal P$ и $C_i\cap C_j=\varnothing$ при $i\ne j$.


# Пример: Борелевская сигма-алгебра

**Определение**

Пусть $(X,\tau)$ — топологическое пространство (то есть на $X$ задана система открытых множеств $\tau$).

**Борелевской $\sigma$-алгеброй** (обозначается $\mathcal B(X)$ или просто $\mathcal B$) называется $\sigma$-алгебра, порождённая семейством всех открытых множеств $X$:

$$
\mathcal B(X)=\sigma(\tau)
$$

где $\sigma(\tau)$ означает минимальную $\sigma$-алгебру, содержащую все открытые множества.

Для $X=\mathbb R$ борелевская $\sigma$-алгебра $\mathcal B(\mathbb R)$ — это минимальная $\sigma$-алгебра, содержащая все интервалы $(a,b)$.

В неё входят:

- все открытые и замкнутые множества;
- все счётные объединения и пересечения открытых/замкнутых множеств (множества типа $F_\sigma$ и $G_\delta$);
- все полуинтервалы $[a,b)$, $(a,b]$, лучи и т. д.

При этом $\mathcal B(\mathbb R)$ строго меньше, чем $2^{\mathbb R}$ (множество всех подмножеств), потому что в $\mathbb R$ существуют неизмеримые по Лебегу множества (например, множества Витали), которые не являются борелевскими.

**Пример полукольца, порождающего борелевскую $\sigma$-алгебру на $\mathbb R$**

Рассмотрим множество $\Omega=\mathbb R$ и семейство

$$
\mathcal P=
\{(a,b]\mid-\infty\le a\le b<+\infty\}
\cup\{\varnothing\}
$$

То есть $\mathcal P$ — это семейство всех полуинтервалов вида $(a,b]$ (включая пустое множество), где $a$ может быть равно $-\infty$, а $b$ — конечное число.

**Вывод:** чтобы задать меру на борелевской сигма-алгебре, достаточно задать меру на полуинтервалах.


# Мера Лебега — Стилтьеса

Пусть $F:\mathbb R\to\mathbb R$ — **неубывающая, непрерывная справа и ограниченная функция**.

Тогда мера Лебега — Стилтьеса $\mu_F$ на борелевской $\sigma$-алгебре $\mathcal B(\mathbb R)$ задаётся на полуинтервалах формулой

$$
\mu_F((a,b])=F(b)-F(a),
\qquad
-\infty<a\le b<+\infty
$$

и единственным образом продолжается на $\mathcal B(\mathbb R)$ по теореме Каратеодори.

**Вывод:** чтобы задать распределение, достаточно задать функцию распределения. Вспомним, что такое распределение и функция распределения.


# Случайная величина и распределение

Пусть задано вероятностное пространство $(\Omega,\mathcal F,\mathbb P)$, где:

- $\Omega$ — пространство элементарных исходов;
- $\mathcal F$ — сигма-алгебра событий;
- $\mathbb P$ — вероятность.

**Случайная величина $X$** — это **измеримая функция** $X:\Omega\to\mathbb R$, такая что для любого борелевского множества $B\subset\mathbb R$ выполняется условие

$$
\{\omega:X(\omega)\in B\}\in\mathcal F
$$

Пусть задано вероятностное пространство $(\Omega,\mathcal F,\mathbb P)$ и случайная величина

$$
X:(\Omega,\mathcal F)\to(\mathbb R,\mathcal B(\mathbb R))
$$

где $\mathcal B(\mathbb R)$ — борелевская $\sigma$-алгебра на $\mathbb R$.

**Распределением** (или **законом**) случайной величины $X$ называется вероятностная мера

$$
\mu_X:\mathcal B(\mathbb R)\to[0,1]
$$

определённая для любого борелевского множества $B\in\mathcal B(\mathbb R)$ равенством

$$
\mu_X(B)
=
\mathbb P\left(X^{-1}(B)\right)
=
\mathbb P\left(\{\omega\in\Omega:X(\omega)\in B\}\right)
$$
