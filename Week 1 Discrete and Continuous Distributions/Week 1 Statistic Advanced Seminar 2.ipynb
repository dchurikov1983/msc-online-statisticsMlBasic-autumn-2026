# Задача 1. Измеримость индикатора

Пусть $(X,M,\mu)$ — измеримое пространство, $A\subseteq X$ и $f(x)=\mathbb{I}_A(x)$. Доказать, что $f$ измерима на $X$ тогда и только тогда, когда $A\in M$.

**Решение**

Напомним, что функция $f:X\to\mathbb{R}$ измерима, если для любого борелевского множества $B\subseteq\mathbb{R}$ прообраз $f^{-1}(B)$ принадлежит $M$.

**Необходимость.** Пусть $f=\mathbb{I}_A$ измерима. Множество $\{1\}$ борелевское, поэтому его прообраз измерим. Но $f^{-1}(\{1\})=A$. Следовательно, $A\in M$.

**Достаточность.** Пусть теперь $A\in M$. Тогда и дополнение $A^c=X\setminus A$ принадлежит $M$.

Индикатор принимает только два значения: $0$ и $1$. Поэтому для любого борелевского $B\subseteq\mathbb{R}$ прообраз $f^{-1}(B)$ может быть только одним из четырёх множеств: $\varnothing$, $A$, $A^c$ или $X$. Все они принадлежат $M$.

Значит, $f$ измерима.

Итак, $\mathbb{I}_A$ измерима тогда и только тогда, когда $A\in M$.

# Задача 2. События сравнения случайных величин

Пусть $\xi$ и $\eta$ — случайные величины, определённые на одном и том же вероятностном пространстве $(\Omega,\mathcal{A},\mathbb{P})$. Доказать, что множества

$A=\{\omega\in\Omega:\xi(\omega)<\eta(\omega)\}$,

$B=\{\omega\in\Omega:\xi(\omega)=\eta(\omega)\}$,

$C=\{\omega\in\Omega:\xi(\omega)\leq\eta(\omega)\}$

принадлежат $\mathcal{A}$, то есть являются событиями.

**Решение**

Начнём с события $A=\{\xi<\eta\}$. Если $\xi(\omega)<\eta(\omega)$, то по плотности рациональных чисел между этими двумя значениями найдётся $q\in\mathbb{Q}$ такое, что $\xi(\omega)<q<\eta(\omega)$. Поэтому

$$
A=\bigcup_{q\in\mathbb{Q}}\{\xi<q\}\cap\{\eta>q\}.
$$

Для каждого рационального $q$ множества $\{\xi<q\}$ и $\{\eta>q\}$ измеримы, поскольку $\xi$ и $\eta$ — случайные величины. Множество $\mathbb{Q}$ счётно, поэтому $A$ является счётным объединением измеримых множеств. Следовательно, $A\in\mathcal{A}$.

Аналогично событие $D=\{\eta<\xi\}$ принадлежит $\mathcal{A}$.

Теперь равенство означает, что не выполняется ни одно из двух строгих неравенств. Поэтому $B=(A\cup D)^c\in\mathcal{A}$.

Наконец, $\xi\leq\eta$ эквивалентно отрицанию $\eta<\xi$. Значит, $C=D^c\in\mathcal{A}$.

Таким образом, все три множества $A$, $B$ и $C$ являются событиями.

# Задача 3. Сигма-алгебра, порождённая случайной величиной

Вероятностное пространство $(\Omega,\mathcal{A},P)$ представляет собой отрезок $[0,1]$ с $\sigma$-алгеброй борелевских подмножеств и мерой Лебега. Описать $\sigma$-алгебру, порождённую случайной величиной $\xi$, если:

a) $\xi(\omega)=1/4$ при $\omega\in[0,1/4)$, $\xi(\omega)=1/2$ при $\omega\in[1/4,3/4)$ и $\xi(\omega)=1$ при $\omega\in[3/4,1]$

б) $\xi(\omega)=\omega/2$

в) $\xi(\omega)=1/2$

г) $\xi(\omega)=\sin(\pi\omega)$

**Решение**

## Пункт 1

**Кусочно-постоянная случайная величина**

Обозначим $E_1=[0,1/4)$, $E_2=[1/4,3/4)$ и $E_3=[3/4,1]$. На этих трёх множествах $\xi$ принимает три различных значения: $1/4$, $1/2$ и $1$.

Каждый атом можно получить как прообраз одноточечного борелевского множества: $E_1=\xi^{-1}(\{1/4\})$, $E_2=\xi^{-1}(\{1/2\})$, $E_3=\xi^{-1}(\{1\})$. Поэтому все три множества принадлежат $\sigma(\xi)$.

С другой стороны, любой прообраз $\xi^{-1}(B)$ зависит только от того, какие из трёх значений $1/4$, $1/2$, $1$ входят в $B$. Значит, любой такой прообраз является объединением некоторых из $E_1,E_2,E_3$.

Следовательно, $\sigma(\xi)$ состоит из всех восьми объединений этих атомов:

$$
\sigma(\xi)=\{\varnothing,\,[0,1/4),\,[1/4,3/4),\,[3/4,1],\,[0,3/4),\,[1/4,1],\,[0,1/4)\cup[3/4,1],\,[0,1]\}.
$$

## Пункт 2

**Случайная величина $\xi(\omega)=\omega/2$**

Отображение $\omega\mapsto\omega/2$ является непрерывной биекцией от $[0,1]$ на $[0,1/2]$, причём обратное отображение $x\mapsto2x$ тоже непрерывно.

Так как $\xi$ непрерывна, она борелевски измерима. Поэтому $\sigma(\xi)\subseteq\mathcal{B}([0,1])$.

Покажем обратное включение. Пусть $A\in\mathcal{B}([0,1])$. Множество $A/2=\{x/2:x\in A\}$ борелевское в $[0,1/2]$, поскольку отображение $x\mapsto x/2$ является гомеоморфизмом. При этом $\xi^{-1}(A/2)=A$.

Значит, каждое борелевское подмножество $[0,1]$ принадлежит $\sigma(\xi)$. Поэтому $\sigma(\xi)=\mathcal{B}([0,1])$.

## Пункт 3

**Постоянная случайная величина $\xi(\omega)=1/2$**

Для любого борелевского множества $B\subseteq\mathbb{R}$ прообраз $\xi^{-1}(B)$ равен либо всему $[0,1]$, если $1/2\in B$, либо пустому множеству, если $1/2\notin B$.

Следовательно, $\sigma(\xi)=\{\varnothing,[0,1]\}$.

## Пункт 4

**Случайная величина $\xi(\omega)=\sin(\pi\omega)$**

Функция симметрична относительно точки $1/2$: для любого $\omega\in[0,1]$ выполняется $\xi(\omega)=\xi(1-\omega)$. Поэтому любой прообраз $\xi^{-1}(B)$ вместе с точкой $\omega$ обязательно содержит и отражённую точку $1-\omega$.

Значит, $\sigma(\xi)$ может содержать только борелевские множества, симметричные относительно $1/2$.

Покажем, что других ограничений нет. Пусть $A\subseteq[0,1]$ — борелевское множество и $\omega\in A$ тогда и только тогда, когда $1-\omega\in A$. Рассмотрим $D=A\cap[0,1/2]$.

На $[0,1/2]$ функция $h(\omega)=\sin(\pi\omega)$ строго возрастает и задаёт гомеоморфизм $[0,1/2]$ на $[0,1]$. Поэтому $E=h(D)$ является борелевским множеством. Из симметрии $A$ следует, что $\xi^{-1}(E)=A$.

Итак,

$$
\sigma(\xi)=\{A\in\mathcal{B}([0,1]):\ \omega\in A\ \Longleftrightarrow\ 1-\omega\in A\}.
$$

То есть $\sigma(\xi)$ состоит ровно из борелевских множеств, симметричных относительно точки $1/2$.

# Задача 4. Множество Витали

Является ли множество Витали измеримым по Лебегу?

На отрезке $[0,1]$ введено отношение эквивалентности $x\sim y$ тогда и только тогда, когда $x-y\in\mathbb{Q}$. Из каждого класса эквивалентности выбрано ровно по одному представителю. Полученное множество обозначается через $V$.

**Решение**

Нет, множество Витали не является измеримым по Лебегу.

Рассмотрим счётное множество $Q=\mathbb{Q}\cap[-1,1]$ и для каждого $q\in Q$ сдвиг $V_q=V+q$.

Сначала заметим, что множества $V_q$ попарно не пересекаются. Действительно, если $v_1+q_1=v_2+q_2$, то $v_1-v_2=q_2-q_1\in\mathbb{Q}$. Значит, $v_1$ и $v_2$ принадлежат одному классу эквивалентности. Но в $V$ выбран ровно один представитель каждого класса, поэтому $v_1=v_2$, а затем и $q_1=q_2$.

Теперь возьмём произвольный $x\in[0,1]$. В его классе эквивалентности есть единственный представитель $v\in V$. Тогда $x-v\in\mathbb{Q}$ и одновременно $x-v\in[-1,1]$. Следовательно, $x\in V_q$ для некоторого $q\in Q$.

Поэтому

$$
[0,1]\subseteq\bigcup_{q\in Q}V_q\subseteq[-1,2].
$$

Предположим, что $V$ измеримо, и обозначим его меру через $m$. Все сдвиги $V_q$ тогда измеримы и имеют ту же меру $m$.

Если $m=0$, то счётное объединение $\bigcup_{q\in Q}V_q$ имеет меру $0$, но оно содержит $[0,1]$, мера которого равна $1$. Получаем противоречие.

Если $m>0$, то из попарной непересекаемости сдвигов и счётной аддитивности следует, что мера их объединения бесконечна. Но это объединение содержится в $[-1,2]$, имеющем конечную меру $3$. Снова противоречие.

Оба варианта невозможны. Следовательно, множество Витали не измеримо по Лебегу.

# Задача 5. Неизмеримая функция с измеримыми уровнями

Пусть $\mu$ — классическая мера Лебега на $[0,1]$. Построить такую **неизмеримую** функцию $f:[0,1]\to\mathbb{R}$, что для любого $c\in\mathbb{R}$ множество $f^{-1}(\{c\})$ измеримо.

**Решение**

Возьмём неизмеримое множество Витали $V\subseteq[0,1]$.

Определим функцию следующим образом: положим $f(x)=x$ для $x\in V$ и $f(x)=x+2$ для $x\in[0,1]\setminus V$.

Покажем сначала, что все множества уровня измеримы. На $V$ значения функции лежат в $[0,1]$, а на дополнении к $V$ — в $[2,3]$. Эти диапазоны не пересекаются. Кроме того, на каждой из двух частей функция инъективна.

Поэтому для любого $c\in\mathbb{R}$ множество $f^{-1}(\{c\})$ содержит не более одной точки. То есть оно либо пусто, либо является одноточечным множеством. Оба варианта измеримы по Лебегу.

Теперь проверим, что сама функция не измерима. Множество $[0,1]$ является борелевским, но

$$
f^{-1}([0,1])=V.
$$

Прообраз борелевского множества оказался неизмеримым. Следовательно, $f$ не является измеримой функцией.

Таким образом, построенная функция неизмерима, хотя каждый её уровень $f^{-1}(\{c\})$ измерим.
