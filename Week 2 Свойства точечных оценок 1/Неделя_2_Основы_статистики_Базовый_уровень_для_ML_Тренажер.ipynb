## Задача 3. Доля посетителей, оставляющих чаевые, и ЦПТ

Сеть ресторанов хочет узнать, какая доля клиентов оставляет чаевые.

Нужно найти:

1. вероятность того, что в выборке из $n=10\,000$ посетителей чаевые оставят от $8\,000$ до $9\,000$ посетителей, если каждый посетитель оставляет чаевые с вероятностью $p=0.89$;
2. минимальное количество наблюдений, чтобы оценить неизвестную вероятность $p$ с точностью не хуже $\pm 0.001$ с вероятностью не меньше $95\%$, если известно, что $p\leq 0.9$.

Для решения воспользуйтесь теоремой Муавра-Лапласа.

**Решение**



### Пункт 1

Пусть $\hat p$ — выборочная доля посетителей, которые оставляют чаевые. По ЦПТ:

$$
\hat p\approx N\left(p,\frac{p(1-p)}{n}\right).
$$

Условие «от $8\,000$ до $9\,000$ из $10\,000$» означает:

$$
0.8\leq \hat p\leq 0.9.
$$

Поэтому:

$$
\mathbb{P}(0.8\leq \hat p\leq 0.9)\approx 0.9993.
$$

```python
from scipy.stats import norm

p = 0.89
n = 10000

std_hat = (p * (1 - p) / n)**.5
prob = norm.cdf(0.9, loc=p, scale=std_hat) - norm.cdf(0.8, loc=p, scale=std_hat)
prob
```

### Пункт 2

Нужно, чтобы:

$$
\mathbb{P}(|\hat p-p|\leq 0.001)\geq 0.95.
$$

По ЦПТ:

$$
\hat p-p\approx N\left(0,\frac{p(1-p)}{n}\right).
$$

Для центрального интервала вероятности $0.95$ используется квантиль:

$$
z_{0.975}\approx 1.96.
$$

Значит, нужно потребовать:

$$
\frac{0.001}{\sqrt{p(1-p)/n}}\geq z_{0.975}.
$$

Отсюда:

$$
n\geq \left(\frac{z_{0.975}}{0.001}\right)^2p(1-p).
$$

Так как $p\leq 0.9$, максимум $p(1-p)$ на отрезке $[0,0.9]$ достигается при $p=0.5$ и равен $0.25$.

Поэтому:

$$
n\geq \left(\frac{1.96}{0.001}\right)^2\cdot 0.25\approx 960\,365.
$$

Минимально нужно примерно $960\,365$ наблюдений.
